## Funções e Polinômios
1. Função:
    Associanção dos elementos de A em B: um valor de A (imagem) não pode está lidado a dois valores de B (contradomínio);
    * 1º Grau: f(x) = 2x + c
    * 2º Grau: f(x) = (ax)^2 + bx + c; 
2. Polonômios: 
    Expressões algébricas formadas por coeficiêntes e vatiáveis com números desconhecidos.
    * Monômios - Possui aoenas um termo
    * Binômios - Possui apenas 2 termos 
    * grau do polinômio determinado pela maior somas dos expoentes
    *Operações com polinômios: Soma ou subtrai os termos iguais e adiciona e troca o sinal dos substraídos

```python
# 1. Funções - Introdução
    #Associanção dos elementos de A em B: um valor de A (imagem) 
    # não pode está lidado a dois valores de B (contradomínio);

dominio_A = [1, 2, 3, 4]
contradominio_B = [1, 2, 3, 4, 5, 6, 7, 8]
imagem = [2, 4, 6, 8] # f(x) = 2x
    # Funções 
    #1. Projetora - Todo elemento de B é imagem de A; 
    #2. Injetora - Pode existir elementos de B sem correspodência em A; 
    #3. Bijetora - Os conjuntos apresentam o mesmo número de elementos relacionados

print(f'Conjunto A - Imagem: {dominio_A}')
print(f'Conjunto b - contradominio:  {contradominio_B}')
print(f'Função: f(x) = 2x:  {imagem}')
```

# Observações: 

f(x) = (ax)^2 + bx + c

* Cria uma parábola e o termo elevado ao quadrado precisa ser diferente de 0, pois vira uma função de 1º grau;
* Se o termo elevado ao quadrado (a) fo maior que 0 a concavidade é pra cima, senão para abaixo; 
* se o segundo termo (b) for maior de 0, a parábola o deslocamento está à direita, senão à esquerda;
* se o segundo termo (b) for igual a 0, a parábola toca o eixo y no ponto mínimo à direita;
* o termo indentente (C) se x - 0 ele intecpta o eixo y.

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Definir os valores de X (intervalo de -10 a 10 com 400 pontos para suavizar a curva)
x = np.linspace(-10, 10, 400)

# 2. Definir as funções (Polinômios)
# Exemplo 1: Função do 1º grau -> f(x) = 2x + 3
# Cria um reta e o X precisa ser diferente de 0 
y_primeiro = 2 * x + 3

# Exemplo 2: Função do 2º grau (Parábola) -> f(x) = x^2 - 4x - 5
y_segundo = x**2 - 4 * x - 5

# Exemplo 3: Polinômios diferentes 
y = 3
y_terceiro = x**3 + 4*x*y +2

# 3. Criar a plotagem
plt.figure(figsize=(8, 5))

# Plotando a reta do 1º grau
plt.plot(x, y_primeiro, label='1º Grau: f(x) = 2x + 3', color='blue')

# Plotando a parábola do 2º grau
plt.plot(x, y_segundo, label='2º Grau: f(x) = x² - 4x - 5', color='red')

plt.plot(x, y_terceiro, label= 'Outros polinônimos com outras variáveis: f(x) =  x3 + 4xy + 2 ', color="orange")

# 4. Personalizações do Gráfico
plt.axhline(0, color='black', linewidth=1, linestyle='--')  # Eixo X
plt.axvline(0, color='black', linewidth=1, linestyle='--')  # Eixo Y
plt.title('Gráficos de Funções Polinomiais')
plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend()

# Exibir o gráfico
plt.show()
```

```python
# Raízes dos polinômios

# Equações do 1º Grau

coeficientes = [3, 4] # Cieficiente a e b = 3x - 4 

# Encontrar as raízes
raizes = np.roots(coeficientes)

print(f'raizes: {raizes}')

```

```python
# Raízes dos polinômios

# Equações do 2º Grau 
# Coeficientes: [a, b, c] para ax² + bx + c = 0
coeficientes = [1, -4, -5]

# Encontrar as raízes
raizes = np.roots(coeficientes)

print("As raízes são x e y:", raizes)
# Saída esperada: [ 5. -1.] (ou seja, x = 5 e x = -1)
```

```python
# Raizes dos polinômios
coeficientes = [-1, 5, 6] # para ax² + bxy + c = 0

raizes = np.roots(coeficientes)

print("As raízes são: x e y", raizes)
```
