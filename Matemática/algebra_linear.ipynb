```python
# Operações com matrizes

matriz_a = np.array([[ 1, 4, 2], [ -2, 1, 0]])
matriz_b = np.array([[ 4, -3, 1], [ 5, 0, 2]])
matriz_c = np.array([[ 2, 0, 3], [ 3, 2, 4]])

print('Operações com matrizes')
# a + b = b = a
print(f'1. Comutativa: \n { matriz_a + matriz_b}')
```

```python
print(f'2. Comutativa: \n { (matriz_a + matriz_b)* matriz_c}')
# (a+b) + c = a (b+c)
```

```python
print(f'3. Elemento oposto:\n { matriz_a +(-matriz_a) }')
# a + (-a) = -a + a = 0
```

```python
print(f'Elemento: \n { matriz_a + 0 }')
# a + 0 = 0 + a = a
```

```python
print(f'1. Comutativa: \n { matriz_a - matriz_b}')
```

```python
matriz_aT = matriz_a.T
print(f'Matriz Transposta: \n {matriz_aT}')
```

```python
# Multiplicação de Matrizes
# O números de colunas A precisar ser igual ao número de linhas de B

# 1. Criando a matriz A com formato (2, 3)
matriz_A = np.array([
    [4, 5, 6]
])

# 2. Criando a transposta com formato (3, 2)
matriz_aT = matriz_A.T

# 3. EXIBINDO AS MATRIZES
print(f'Matriz A (shape {matriz_A.shape}):\n{matriz_A}\n')
print(f'Matriz Transposta (shape {matriz_aT.shape}):\n{matriz_aT}\n')

# -------------------------------------------------------------
# O SEGREDO ESTÁ AQUI: Use obrigatoriamente o símbolo '@'
# -------------------------------------------------------------
resultado = matriz_A @ matriz_aT

print(f'Resultado da Multiplicação Matricial (shape {resultado.shape}):\n{resultado}')
```

```python
# Propriedades da multiplicação (Erro)

matriz_a = np.array([[ 2, 4, 5], [ 1, -2, 0 ]])
matriz_b = np.array([[ 3, 0], [ 3, 0 ], [1, -3]])
matriz_c = np.array([[ 1, 2, 3], [ 0, -1, 2 ]])

print(f'Matriz A: \n {matriz_a}')

print(f'Matriz B: \n {matriz_b}')

print(f'Matriz C: \n {matriz_c}')
```

```python
# Não é comulativa a x b diferente de b x a

print(f'Multiplicação de A e B: \n {matriz_a @ matriz_b}')

print(f'\n Multiplicação de B e A: \n {matriz_b @ matriz_a}')
```

```python
# Assossiativa: (a x b)x c = a x(b x c) - São iguais

print(f'Multiplicação de (A x B) x c: \n {(matriz_a @ matriz_b) @ matriz_c}')

print(f'Multiplicação de A x (B x C): \n {matriz_a @ (matriz_b @ matriz_c)}')

```

```python
# X x ( a + c ) = b x a + c x a
# A correção é feita usando matrizes com dimensões compatíveis para a adição.
# `matriz_a` e `matriz_c` têm ambos o formato (2,3), o que permite a adição.
# `matriz_x` também será convertida para um array NumPy com o formato (2,3).

matriz_x_np = np.array([[2, -3, 0], [1, 0, 2]])

# Realizando a operação com matrizes de dimensões compatíveis
# O operador '*' realiza multiplicação elemento a elemento.
# Usamos 'matriz_a + matriz_c' pois ambos têm dimensão (2,3).
print(f'Multiplicação elemento a elemento de X com (A + C): \n {matriz_x_np * (matriz_a + matriz_c)}')
```

```python
# Elemento a x ln = ln. a = a

print(f'Elemento neutro: \n {matriz_a * 1}')
```

```python
# Determinantes

# 1. Gerando a matriz quadrada (2x2) que você calculou antes
matriz_quadrada = matriz_a @ matriz_a.T

# 2. Calculando o determinante
determinante = np.linalg.det(matriz_quadrada)

print(f'Matriz Quadrada:\n{matriz_quadrada}\n')
print(f'Determinante da Matriz: {determinante:.2f}')
```
