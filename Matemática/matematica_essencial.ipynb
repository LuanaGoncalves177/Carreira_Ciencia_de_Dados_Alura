```python
# 6. Logaritmo - operação inversa da exponenciação
# O logaritmo é a operação inversa da exponenciação que responde à pergunta:
#  "A que potência devemos elevar a base para obter um determinado número?".

#b ^n = a 

# Logaritmos de mesma base: log_b(b^n) = n

# Definição de Logaritmo: Número n é o logaritmo de a na base b se, 
# e somente se, b elevado a n é igual a a.
# log_b(a) = n

print("Exemplo de logaritmo:")

b = 3,
a = 2187
#log_3(2187) = 7, pois 3^7 = 2187  

print(f"log_{b}({a}) = {7}, pois {b}^{7} = {a}")

#Produto de logaritmos: log_b(m * n) = log_b(m) + log_b(n) 
# mesma regra do produto de potências: b^m * b^n = b^(m+n)
a = 9
b = 27
#log_3(27) = log_3(3 * 9) = (log_3(3) = 1 + log_3(9) = 1 + 2 = 3)
print(f"Produto de logaritmos: log_{a}({a*b}) = log_{a}({a}) + log_{a}({b}) ,ou seja, log_{a}({a*b}) = {a} x {b} = {a*b}")

#Quaciente de logaritmos: log_b(m / n) = log_b(m) - log_b(n)
# mesma regra do quociente de potências: b^m / b^n = b^(m-n)
a = 81
b = 9
#log_3(81) = log_3(9 * 9) = (log_3(9) = 2 - log_3(9) = 2 - 2 = 0)
print(f"Quociente de logaritmos: log_{a}({a/b}) = log_{a}({a}) - log_{a}({b})")

# 6. Logaritmo - operação inversa da exponenciação
from math import log

# Potência de logaritmo
# log_a(p^n) = n * log_a(p)
a = 10
p = 1000
n =3

result = n * (math.log(p, a))
print(f"Resultado potência de logaritmo: {result}")
```

```python
# 5. Radição - operação inversa da potenciação
# Propriedades da radiciação:

# Divisão de radicais: √a / √b = √(a/b)
a = 100
b = 25
radical_divisao = (a ** 0.5) / (b ** 0.5) # = 100 elevado a 0.5 = 10, e 25 elevado a 0.5 = 5, 
#então 10 / 5 = 2  ou raiz quadrada de 100 dividido pela raiz quadrada de 25 = raiz quadrada de 4 = 2
print("Divisão de radicais:", radical_divisao)


# Exponenciação de radicais: (√a)^n = a^(n/2)
a = 16
n = 3
m = 2

# Jeito 1: Seguindo o passo a passo da imagem (potência primeiro, depois a raiz cúbica)
resultado_1 = (8 ** 2) ** (1/3)

# Jeito 2: Usando direto a propriedade do expoente fracionário (2/3)
resultado_2 = 8 ** (2/3)

print("Resultado:", resultado_1, "ou", resultado_2)

resultado = round(8 ** (2/3))
print("Arrendodamento: ", resultado)  # Retorna 4
```

```python
# 5. Radição - operação inversa da potenciação
# A radiciação é a operação inversa da potenciação. 

#x elevado a 0.5 = √x
#x elevado a 1/n = n√x
#X e n são números reais, e n é diferente de 0.

x = 16 
n = 2 
a = x ** (1/n) # = 16 elevado a 1/2 = 4
print("Radiciação: ", a) 

# Propriedade da radiciação: a raiz quadrada de um número 
# ao quadrado é igual ao próprio número.

a = 100
b = 49
n = 2
result1 = (a ** n) ** (1/n) # = 100 elevado a 2 = 10000, e depois a raiz quadrada de 10000 = 100
result2 = (b ** n) ** (1/n) # = 49 elevado a 2 = 2401, e depois a raiz quadrada de 2401 = 49
print("Propriedade da radiciação: ", result1, result2)  

(a * b) ** (1/n) == (a ** (1/n)) * (b ** (1/n)) # = 100 * 49 = 4900, e depois a raiz quadrada de 4900 = 70
print("Propriedade da radiciação: ", (a * b) ** (1/n), (a ** (1/n)) * (b ** (1/n)))
```

```python
# 4. Potenciação - multiplicação repetida

# Casos Especiais:
# - Qualquer número elevado a 0 é igual a 1.
x = 2 
r = x ** 0
print("Qualquer número elevado a 0 é igual a 1:", r == 1)

# divisão de potências com a mesma base: subtração dos expoentes
x = 2
y = 3
r = x ** y / x ** 2 # 2*3 / 2*2 = 2^(3-2) = 2^1 = 2 
print("Divisão de potências com a mesma base:", r)
print("Divisão de potências com a mesma base:", r == x ** (y - 2))

#potenciação de números negativos: o resultado será positivo 
#se o expoente for par e negativo se o expoente for ímpar.

x = -2
y = 4
w = 3
r = x ** y # (-2)^4 = 16
print("Potência de um número negativo (expoente par 4):", r)
r2 = x ** w # (-2)^3 = -8
print("Potência de um número negativo (expoente ímpar 3):", r2)

# Notação científica: é uma forma de representar números muito grandes 
# ou muito pequenos, utilizando potências de 10.

x = 1.23e4 # 1.23 * 10^4 = 12300
print("Notação científica:", x)

y = 5.67e-3 # 5.67 * 10^-3 = 0.00567
print("Notação científica:", y)
```

```python
# 4. Potenciação - multiplicação repetida

# Propriedades da potenciação: Multiplicação e divisão de potências com a mesma base

# Multiplicação de potências com a mesma base: a^m * a^n = a^(m+n)
a = 2
n = 3 
m = 4
result1 = a**n * a**m
print(f"Multiplicação: {a}^{n} * {a}^{m} = {result1}")
result2 = a**(n+m)
print(f"Multiplicação: {a}^{n+m} = {result2}")
print("Propriedade da multiplicação de potências com a mesma base:", result1 == result2)

# Divisão de potências com a mesma base: a^m / a^n = a^(m-n)
result2 = a**m / a**n
print(f"Divisão: {a}^{m} / {a}^{n} = {result2}")
result3 = a**(m-n)
print(f"Divisão: {a}^{m-n} = {result3}")
print("Propriedade da divisão de potências com a mesma base:", result2 == result3)

# 4. Potenciação - multiplicação repetida
#comutatividade da multiplicação
x = 3 
y = 4
print(f"Comutatividade da multiplicação: {x} * {y} = {x * y}")
print(f"Comutatividade da multiplicação: {y} * {x} = {y * x}")
```

```python
#2. Frações e números decimais;

from fractions import Fraction
#Propriedade da adição e subtração de frações e números decimais

#Soma e subtração de frações e números decimais

# Exemplo de frações
frac1 = Fraction(1, 2)  # 1/2
frac2 = Fraction(1, 3)  # 1/3

# Soma de frações (1/2 + 1/3 = 5/6) -> 1 x 3 + 1 x 2 / 2 x 3 = 5/6
soma = frac1 + frac2
print("Soma de frações:", soma)

# Subtração de frações (1/2 - 1/3 = 1/6) -> 1 x 3 - 1 x 2 / 2 x 3 = 1/6 
subtracao = frac1 - frac2
print("Subtração de frações:", subtracao)

# Números decimais
num1 = 0.5
num2 = 0.3

# Soma de números decimais
soma_dec = num1 + num2
print("Soma de números decimais:", soma_dec)

# Subtração de números decimais
subtracao_dec = num1 - num2
print("Subtração de números decimais:", subtracao_dec)
```
