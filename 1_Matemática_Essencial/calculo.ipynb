    *lim(x-> a) f(x) = L
    *Limite: Determinar o comportamento de uma função à medida que ela se aproxima de alguns valores, sempre relacionado os pontos x e y; 
    * Limites laterais 
        Aproximação pela esquerda:  lim(x-> a-) f(x) = L
        Aproximação pela esquerda:  lim(x-> a+) f(x) = L

    * continuidade de uma função: função sem quebra ou salto: Condições: 
        * f(x) está fefinida 
        * lim(x_> a) f(x) existe
        * lim(x_> a) f(x) = f(a)
    *Derivada: Taxa de variação de um função em um determinado ponto. 
    * f'(x) = (df/dx)(x)
        * Define o cresimento e decrescimento de uma função
        # Deltas
            Delta_x = x1 - x0 
            Delta_y = y1 - y0
            Taxa de variação
                Delta_y/Delta_x = (f(x1) - f(x0))/(x1 - x0)
        Delta X tente a zero(0)
            f'(x) = lim(x->0)(f(x)-f(x0)/(x-x0)) 
            = lim(Delta -> 0 (f(x0 + delta_x)- f(x0) delta_x))
            Função 1º Grau:  é zero para qualquer valor de x
            Função 2º Frau: é o a: o próprio coeficiênte
            Função polinomial: aplicar detivada polinomial termo a termo
        Ponto mínimo e máximo de uma função
    Gradiente: é um vetor que indica o sentido e a direção na qual, por deslocamento a partir de um ponto espercífico. 
        * derivdas parciais em relação a Y e a X. 

```python
import sympy as sp

# 1. Definir a variável simbólica
x = sp.Symbol('x')

# 2. Definir uma função (ex: f(x) = x³ - 4x² + 5)
f = x**3 - 4 * x**2 + 5
print('Função: Ex: f(x) = x³ - 4x² + 5): ',
{f})

# 3. Calcular a Derivada de f(x) em relação a x
derivada = sp.diff(f, x)
print('Derivada:', derivada)  # Saída: 3*x**2 - 8*x

# 4. Calcular a Integral Indefinida
integral = sp.integrate(f, x)
print('Integral:', integral)  # Saída: x**4/4 - 4*x**3/3 + 5*x
```

```python
# 1. Definir as variáveis simbólicas
x, y = sp.symbols('x y')

# 2. Definir uma função com várias variáveis (ex: f(x, y) = x³*y² + 4*x*y)
f = x**3 * y**2 + 4 * x * y

# 3. Calcular a Derivada Parcial em relação a x (trata y como constante)
derivada_x = sp.diff(f, x)
print("Derivada parcial em relação a x:", derivada_x)
# Saída esperada: 3*x**2*y**2 + 4*y

# 4. Calcular a Derivada Parcial em relação a y (trata x como constante)
derivada_y = sp.diff(f, y)
print("Derivada parcial em relação a y:", derivada_y)
# Saída esperada: 2*x**3*y + 4*x
```
