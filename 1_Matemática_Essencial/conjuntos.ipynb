```python
#4. Lógica proporcional: utiliza para establecer relações entre diferentes variáveis e  determinar como uma mudança em uma variável afeta outra. Por exemplo, se a quantidade de horas de estudo aumenta, a nota do aluno tende a melhorar proporcionalmente.

#proporção -> uma declaração que pode ser verdadeira ou falsa
#P = V ou F

#4.1. Comectivos lógicos: são operadores que conectam proposições para formar novas proposições. Os principais conectivos lógicos são:
#Disjunção (OR): A proposição "A ou B" é verdadeira se pelo menos uma das proposições A ou B for verdadeira. -> V
#Conjunção (AND): A proposição "A e B" é verdadeira apenas se ambas as proposições A e B forem verdadeiras. -> ^
#Negação (NOT): A proposição "não A" é verdadeira se a proposição A for falsa, e falsa se a proposição A for verdadeira.   
# condição (IF...THEN): A proposição "Se A, então B" é verdadeira se sempre que A for verdadeira, B também for verdadeira. Se A for falsa, a proposição é considerada verdadeira independentemente do valor de B.   
# bicondicional (IF AND ONLY IF): A proposição "A se e somente se B" é verdadeira se A e B tiverem o mesmo valor de verdade, ou seja, ambas forem verdadeiras ou ambas forem falsas.   
# #4.2. Tabela verdade: é uma ferramenta usada para representar todas as combinações possíveis de valores de verdade das proposições envolvidas em uma expressão lógica. Ela ajuda a determinar o valor de verdade da expressão como um todo com base nos valores de verdade das proposições individuais.
# #4.3. Leis da lógica: são regras que descrevem como as proposições podem ser manipuladas e combinadas. Algumas das leis mais importantes incluem:
# Lei da identidade: A proposição A é equivalente a A.  
# Lei da não contradição: A proposição A e sua negação ¬A não podem ser verdadeiras ao mesmo tempo.
# Lei do terceiro excluído: A proposição A ou sua negação ¬A deve ser verdadeira, ou seja, não há uma terceira opção. 
# Lei da dupla negação: A proposição ¬(¬A) é equivalente a A.
# Lei de De Morgan: As leis de De Morgan descrevem como a negação de uma conjunção ou disjunção pode ser expressa em termos das negações das proposições individuais. Elas afirmam que ¬(A ∧ B) é equivalente a ¬A ∨ ¬B, e ¬(A ∨ B) é equivalente a ¬A ∧ ¬B.
# lei da distributividade: A proposição A ∧ (B ∨ C) é equivalente a (A ∧ B) ∨ (A ∧ C), e A ∨ (B ∧ C) é equivalente a (A ∨ B) ∧ (A ∨ C). 
# #4.4. Argumentos lógicos: são conjuntos de proposições que visam estabelecer a validade de uma conclusão com base em premissas. Um argumento é considerado válido se a conclusão segue logicamente das premissas, ou seja, se as premissas forem verdadeiras, a conclusão também será verdadeira. A validade de um argumento pode ser avaliada por meio de regras de inferência e análise lógica.
# #4.5. Inferência lógica: é o processo de derivar conclusões a partir de premissas utilizando regras de inferência válidas. As regras de inferência permitem que se faça deduções lógicas com base nas proposições existentes, garantindo que as conclusões obtidas sejam consistentes com as premissas.
# #4.6. Quantificadores: são símbolos utilizados para expressar a quantidade de elementos em um conjunto que satisfazem uma determinada propriedade. Os principais quantificadores são:
# Quantificador universal (∀): indica que uma propriedade é verdadeira para todos os elementos de um conjunto. Por exemplo, "Para todo x, se x é um número par, então x é divisível por 2."
# Quantificador existencial (∃): indica que existe pelo menos um elemento em um conjunto
# que satisfaz uma determinada propriedade. Por exemplo, "Existe um x tal que x é um número primo."

```

```python
# 3. Fundamentos matemáticos
# 3.1. Operações com conjuntos 

# 3.1.1. União de conjuntos A U B - Todos os elementos de A e B
uniao = numeros.union(pares)
print(uniao)  # {1, 2, 3, 4, 5}

# 3.1.2. Interseção de conjuntos A ∩ B - Elementos que estão em ambos os conjuntos
intersecao = numeros.intersection(pares)
print(intersecao)  # {2, 4}

#3.1.3. Diferença de conjuntos A - B - Elementos que estão em A mas não em B
diferenca = numeros.difference(pares)  
print(diferenca)  # {1, 3, 5}

#3.1.4. Complemento de conjuntos A' - Elementos que não estão em A
complemento = set(range(1, 6)).difference(numeros)  # {6, 7, 8, 9, 10}
print(complemento)  # {6, 7, 8, 9, 10}
```

```python
# 1. Conjuntos númericos 

numeros = {1, 2, 3, 4, 5}
pares = {2, 4,}
impares = {1, 3, 5, 7, 9}

# 2. Operações básicas com conjuntos
# 2.1 Pertinencia: verificar si un elemento pertenece a un conjunto
print(3 in numeros)  # True
print(3 in pares)    # False
print(3 in impares)  # True

# 2.2 Subconuntos: verificar si un conjunto es subconjunto de otro
print(pares.issubset(numeros))  # True
print(impares.issubset(numeros))  # False
```
