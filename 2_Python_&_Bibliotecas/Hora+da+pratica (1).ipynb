#### Questão 9
Em uma eleição para gerência em uma empresa com 20 funcionários, existem quatro candidatos. Escreva um programa que calcule o vencedor da eleição. A votação ocorreu da seguinte maneira:
- Cada funcionário votou em um dos quatro candidatos (representados pelos números 1, 2, 3 e 4).
- Também foram contabilizados os votos nulos (representado pelo número 5) e os votos em branco (representado pelo número 6).

Ao final da votação, o programa deve exibir o total de votos para cada candidato, o número de votos nulos e o número de votos em branco. Além disso, deve calcular e exibir a porcentagem de votos nulos em relação ao total de votos e a porcentagem de votos em branco em relação ao total de votos.

````python
votos = [0, 0, 0, 0, 0, 0]
brancos = 0
nulos = 0
funcionarios = 21

for i in range(1, funcionarios):

    voto = int(input(f"Candidato {i}: "))
    if voto == 1:
        votos[0] += 1
    elif voto == 2:
        votos[1] += 1
    elif voto == 3:
        votos[2] += 1
    elif voto == 4:
        votos[3] += 1
    elif voto == 5:
        nulos += 1
    elif voto == 6:
        brancos += 1
    else:
        print("Voto inválido")

print("\nTotal de votos para cada candidato:")
print(f"Candidato 1: {votos[0]} votos")
print(f"Candidato 2: {votos[1]} votos")
print(f"Candidato 3: {votos[2]} votos")
print(f"Candidato 4: {votos[3]} votos")
print(f"Votos nulos: {nulos} votos")
print(f"Votos em branco: {brancos} votos")
````

````python
limite = int(input("Digite um número: "))

primos = []
for num in range(2, limite + 1):
  is_primo = True
  for i in range(2, int(num**0.5) + 1):
    if num % i == 0:
      is_primo = False
      break
  if is_primo:
    primos.append(num)

print(f"Os números primos até {limite} são: {primos}")
````

Tendo esses valores, faça um código que gere uma lista contendo o percentual de crescimento de bactérias por dia, comparando o número de bactérias em cada dia com o número de bactérias do dia anterior. Dica: para calcular o percentual de crescimento usamos a seguinte equação: `100 * (amostra_atual - amostra_passada) / (amostra_passada)`.

````python
Col_bacteria = [1.2, 2.1, 3.3, 5.0, 7.8, 11.3, 16.6, 25.1, 37.8, 56.9]

percentual_crescimento = []
for i in range(1, len(Col_bacteria)):
    crescimento = 100 * (Col_bacteria[i] - Col_bacteria[i-1]) / Col_bacteria[i-1]
    percentual_crescimento.append(crescimento)

print(percentual_crescimento)
````

#### Questão 15
O setor de RH da sua empresa te pediu uma ajuda para analisar as idades dos funcionários de 4 setores da empresa. Para isso, ele te forneceu os seguintes dados:
```
{'Setor A': [22, 26, 30, 30, 35, 38, 40, 56, 57, 65],
 'Setor B': [22, 24, 26, 33, 41, 49, 50, 54, 60, 64],
 'Setor C': [23, 26, 26, 29, 34, 35, 36, 41, 52, 56],
 'Setor D': [19, 20, 25, 27, 34, 39, 42, 44, 50, 65]}
```
Sabendo que cada setor tem 10 funcionários, construa um código que calcule a média de idade de cada setor, a idade média geral entre todos os setores e quantas pessoas estão acima da idade média geral.


````python
setores = {'Setor A': [22, 26, 30, 30, 35, 38, 40, 56, 57, 65],
 'Setor B': [22, 24, 26, 33, 41, 49, 50, 54, 60, 64],
 'Setor C': [23, 26, 26, 29, 34, 35, 36, 41, 52, 56],
 'Setor D': [19, 20, 25, 27, 34, 39, 42, 44, 50, 65]}

# Calcula a média de idade de cada setor
for setor, idades in setores.items():
    media_idade = sum(idades) / len(idades)
    print(f"A média de idade do setor {setor} é: {media_idade:.2f}")

# Calcula a idade média geral entre todos os setores
total_idades_geral = sum(sum(idades) for idades in setores.values())
total_pessoas_geral = sum(len(idades) for idades in setores.values())
media_idade_geral = total_idades_geral / total_pessoas_geral
print(f"\nA idade média geral entre os setores é: {media_idade_geral:.2f}")

# Calcula o número de pessoas acima da idade média geral
pessoas_acima_media_geral = 0
for idades in setores.values():
    for idade in idades:
        if idade > media_idade_geral:
            pessoas_acima_media_geral += 1

print(f"O número de pessoas acima da idade média geral é: {pessoas_acima_media_geral}")
````
