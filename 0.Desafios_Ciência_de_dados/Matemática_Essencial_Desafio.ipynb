
1. Conjuntos e Lógica

Em uma equipe de tecnologia, um grupo de 30 profissionais utiliza ferramentas de análise. Sabe-se que 18 usam Python, 15 usam R, e 6 usam ambas as linguagens.

Quantos profissionais dessa equipe não usam nenhuma das duas linguagens?

```python
# Variáveis
equipe = 30
python_R = 6
python = 18 - python_R
Ling_R = 15 - python_R

outros = equipe - python - Ling_R - python_R
print(f'Não programam e nenhuma linguagem é: {outros}')
```

2. Álgebra Linear

Dada a matriz $A = \begin{bmatrix} 2 & 1 \\ 0 & 3 \end{bmatrix}$ e o vetor $v = \begin{bmatrix} 1 \\ 2 \end{bmatrix}$, qual é o resultado da multiplicação da matriz pelo vetor ($A \cdot v$)?

```python
# Variáveis
matrix_A = np.array([[2, 1],[1, 3]])
vetor_V = np.array([[1],[2]])

# Multiplicação

print(f'Matrix A:\n {matrix_A}')
print(f"Vetor V: \n {vetor_V}")
print(f'Multiplicação da Matrix A pelo vetor V:\n {matrix_A @ vetor_V}')
```

3. Funções e Polinômios

Qual é o volume de dados $x$ que minimiza o custo de processamento dessa empresa?

4. Cálculo (Derivadas e Otimização)

Dada a função de custo $f(w) = 3w^2 - 12w + 5$, qual é o valor de $w$ que zera a derivada da função ($f'(w) = 0$), ponto que representa o mínimo da função em um processo de gradiente?


```python
# Função de custo f(w) = 3w^2 - 12w + 5

# A derivada da função é f'(w) = 6w - 12
# Para encontrar o mínimo, igualamos a derivada a zero: 6w - 12 = 0

# Resolvendo para w:
w = 12 / 6

print(f'O valor de w que zera a derivada da função é: {w}')
```

5. Matemática Essencial (Potenciação e Logaritmos)

Simplificando a expressão com logaritmos e potências que você revisou no código: sabendo que $\log_3(2187) = 7$ (porque $3^7 = 2187$), qual é o resultado da operação de potenciação $2^3 \cdot 2^4$ utilizando as propriedades de bases iguais?



```python
# Variáveis:

pot1 = 2**3
pot2 = 2**4
potR = 2**(3+4)

print(f'Primeira Potência 2^3: {pot1}')
print(f'Segunda Potência 2^4: {pot2}')
print(f'Potência resultante 2^7: {potR}')
```

```python
import math

# Operação de Logaritmo: log base 3 de 2187
log_result = math.log(2187, 3)
print(f'log_3(2187) = {log_result}')

# Operação de Potenciação: 2^3 * 2^4 usando propriedades de bases iguais
# Isso é equivalente a 2^(3+4) = 2^7
pow_result_base3 = math.pow(2, 3)
pow_result_base4 = math.pow(2, 4)
pow_result_product = pow_result_base3 * pow_result_base4
print(f'2^3 * 2^4 = {pow_result_product}')

# Ou diretamente 2^(3+4)
pow_result_combined = math.pow(2, 3 + 4)
print(f'2^(3+4) = {pow_result_combined}')
```
